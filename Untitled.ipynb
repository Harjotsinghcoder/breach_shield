```python
# Set up the indexed dataframe and aggregate the state features
df_indexed = df.set_index('Timestamp')

# 1. Flow count per window
flow_count = df_indexed.resample('1min').size().rename('flow_count')

# 2. Aggregate the numeric flow features into summary stats per window
agg_dict = {
    'Flow Duration': ['mean', 'std'],
    'Tot Fwd Pkts': 'sum',
    'Tot Bwd Pkts': 'sum',
    'TotLen Fwd Pkts': 'sum',
    'TotLen Bwd Pkts': 'sum',
    'Flow Byts/s': 'mean',
    'Flow Pkts/s': 'mean',
    'Flow IAT Mean': 'mean',
    'SYN Flag Cnt': 'sum',
    'ACK Flag Cnt': 'sum',
    'RST Flag Cnt': 'sum',
    'FIN Flag Cnt': 'sum',
    'PSH Flag Cnt': 'sum',
    'URG Flag Cnt': 'sum',
    'Dst Port': pd.Series.nunique,
}
state_features = df_indexed.resample('1min').agg(agg_dict)
state_features.columns = ['_'.join(c).strip('_') for c in state_features.columns]
state_features = state_features.rename(columns={'Dst Port_nunique': 'unique_dst_ports'})

# 3. Label signals per window
def label_stats(labels):
    if len(labels) == 0:
        return pd.Series({'dominant_label': 'NoTraffic', 'attack_ratio': np.nan})
    counts = labels.value_counts()
    dominant = counts.idxmax()
    attack_ratio = (labels != 'Benign').mean()
    return pd.Series({'dominant_label': dominant, 'attack_ratio': attack_ratio})

label_info = df_indexed.groupby(pd.Grouper(freq='1min'))['Label'].apply(label_stats).unstack()

# 4. Combine everything
state_df = pd.concat([flow_count, state_features, label_info], axis=1)
state_df['attack_present'] = (state_df['attack_ratio'] > 0).astype(int)

print("State vector shape:", state_df.shape)
print(state_df.head(10))
```

```python
# Combine scaled traffic features (17) + attack_ratio (already 0-1, kept unscaled) = 18-dim state vector
train_full = np.hstack([train_scaled, train_df[['attack_ratio']].values])
test_full = np.hstack([test_scaled, test_df[['attack_ratio']].values])

print("Train full state shape:", train_full.shape)
print("Test full state shape:", test_full.shape)

def make_sequences(data, window_size):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i+window_size])   # 10-minute window (the "given")
        y.append(data[i+window_size])      # the very next minute's state (the "predict")
    return np.array(X), np.array(y)

WINDOW_SIZE = 10
X_train, y_train = make_sequences(train_full, WINDOW_SIZE)
X_test, y_test = make_sequences(test_full, WINDOW_SIZE)

print("\nX_train shape:", X_train.shape, "(samples, window_size, features)")
print("y_train shape:", y_train.shape, "(samples, features)")
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
def weighted_mse(pred, target, attack_weight=5.0):
    weights = torch.ones(pred.shape[1], device=pred.device)
    weights[-1] = attack_weight  # last column is attack_ratio
    return ((pred - target) ** 2 * weights).mean()

model = WorldModel(n_features).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

EPOCHS = 30
for epoch in range(EPOCHS):
    model.train()
    total_loss = 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = weighted_mse(model(xb), yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
    train_loss = total_loss / len(X_train_t)

    model.eval()
    with torch.no_grad():
        test_loss = 0
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            test_loss += weighted_mse(model(xb), yb).item() * xb.size(0)
        test_loss /= len(X_test_t)

    if epoch % 5 == 0 or epoch == EPOCHS - 1:
        print(f"Epoch {epoch:2d} | train wMSE {train_loss:.4f} | test wMSE {test_loss:.4f}")
```

```python
def make_multistep_sequences(data, window_size, horizon):
    X, Y = [], []
    for i in range(len(data) - window_size - horizon + 1):
        X.append(data[i:i+window_size])
        Y.append(data[i+window_size : i+window_size+horizon])
    return np.array(X), np.array(Y)

HORIZON = 5  # train the model to stay accurate 5 steps into its own rollout

X_train_multi, Y_train_multi = make_multistep_sequences(train_full2, WINDOW_SIZE, HORIZON)
X_test_multi, Y_test_multi = make_multistep_sequences(test_full2, WINDOW_SIZE, HORIZON)

print(X_train_multi.shape, Y_train_multi.shape)
print(X_test_multi.shape, Y_test_multi.shape)
```

```python
def train_multistep(model, X, Y, optimizer, horizon, attack_weight=5.0, batch_size=32):
    model.train()
    X_t = torch.tensor(X, dtype=torch.float32)
    Y_t = torch.tensor(Y, dtype=torch.float32)
    loader = DataLoader(TensorDataset(X_t, Y_t), batch_size=batch_size, shuffle=True)

    total_loss = 0
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        window = xb
        loss = 0
        for t in range(horizon):
            pred = model(window)              # predict next state
            target = yb[:, t, :]              # true state at this horizon step
            loss += weighted_mse(pred, target, attack_weight)
            window = torch.cat([window[:, 1:, :], pred.unsqueeze(1)], dim=1)  # feed own prediction back in
        loss = loss / horizon
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
    return total_loss / len(X)


def eval_multistep(model, X, Y, horizon, attack_weight=5.0, batch_size=32):
    model.eval()
    X_t = torch.tensor(X, dtype=torch.float32)
    Y_t = torch.tensor(Y, dtype=torch.float32)
    loader = DataLoader(TensorDataset(X_t, Y_t), batch_size=batch_size, shuffle=False)

    total_loss = 0
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            window = xb
            loss = 0
            for t in range(horizon):
                pred = model(window)
                target = yb[:, t, :]
                loss += weighted_mse(pred, target, attack_weight)
                window = torch.cat([window[:, 1:, :], pred.unsqueeze(1)], dim=1)
            loss = loss / horizon
            total_loss += loss.item() * xb.size(0)
    return total_loss / len(X)

print("Functions defined.")
```
