```python
# Set up the indexed dataframe and aggregate the state features
df_indexed = df.set_index('Timestamp')

# 1. Flow count per window
flow_count = df_indexed.resample('1min').size().rename('flow_count')

# 2. Aggregate the numeric flow features into summary stats per window
agg_dict = {
    'Flow Duration': ['mean', 'std'],
    'Tot Fwd Pkts': 'sum',
    'Tot Bwd Pkts': 'sum',
    'TotLen Fwd Pkts': 'sum',
    'TotLen Bwd Pkts': 'sum',
    'Flow Byts/s': 'mean',
    'Flow Pkts/s': 'mean',
    'Flow IAT Mean': 'mean',
    'SYN Flag Cnt': 'sum',
    'ACK Flag Cnt': 'sum',
    'RST Flag Cnt': 'sum',
    'FIN Flag Cnt': 'sum',
    'PSH Flag Cnt': 'sum',
    'URG Flag Cnt': 'sum',
    'Dst Port': pd.Series.nunique,
}
state_features = df_indexed.resample('1min').agg(agg_dict)
state_features.columns = ['_'.join(c).strip('_') for c in state_features.columns]
state_features = state_features.rename(columns={'Dst Port_nunique': 'unique_dst_ports'})

# 3. Label signals per window
def label_stats(labels):
    if len(labels) == 0:
        return pd.Series({'dominant_label': 'NoTraffic', 'attack_ratio': np.nan})
    counts = labels.value_counts()
    dominant = counts.idxmax()
    attack_ratio = (labels != 'Benign').mean()
    return pd.Series({'dominant_label': dominant, 'attack_ratio': attack_ratio})

label_info = df_indexed.groupby(pd.Grouper(freq='1min'))['Label'].apply(label_stats).unstack()

# 4. Combine everything
state_df = pd.concat([flow_count, state_features, label_info], axis=1)
state_df['attack_present'] = (state_df['attack_ratio'] > 0).astype(int)

print("State vector shape:", state_df.shape)
print(state_df.head(10))
```

```python
feature_cols = [
    'flow_count', 'Flow Duration_mean', 'Flow Duration_std',
    'Tot Fwd Pkts_sum', 'Tot Bwd Pkts_sum', 'TotLen Fwd Pkts_sum', 'TotLen Bwd Pkts_sum',
    'Flow Byts/s_mean', 'Flow Pkts/s_mean', 'Flow IAT Mean_mean',
    'SYN Flag Cnt_sum', 'ACK Flag Cnt_sum', 'RST Flag Cnt_sum', 'FIN Flag Cnt_sum',
    'PSH Flag Cnt_sum', 'URG Flag Cnt_sum', 'unique_dst_ports'
]
target_col = 'attack_ratio'

split_idx = int(len(state_df) * 0.8)
train_df = state_df.iloc[:split_idx]
test_df = state_df.iloc[split_idx:]

print(f"Train: {len(train_df)} minutes ({train_df.index.min()} to {train_df.index.max()})")
print(f"Test:  {len(test_df)} minutes ({test_df.index.min()} to {test_df.index.max()})")

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_df[feature_cols])
test_scaled = scaler.transform(test_df[feature_cols])

print("\nTrain scaled shape:", train_scaled.shape)
print("Test scaled shape:", test_scaled.shape)
```

```python
# Combine scaled traffic features (17) + attack_ratio (already 0-1, kept unscaled) = 18-dim state vector
train_full = np.hstack([train_scaled, train_df[['attack_ratio']].values])
test_full = np.hstack([test_scaled, test_df[['attack_ratio']].values])

print("Train full state shape:", train_full.shape)
print("Test full state shape:", test_full.shape)

def make_sequences(data, window_size):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i+window_size])   # 10-minute window (the "given")
        y.append(data[i+window_size])      # the very next minute's state (the "predict")
    return np.array(X), np.array(y)

WINDOW_SIZE = 10
X_train, y_train = make_sequences(train_full, WINDOW_SIZE)
X_test, y_test = make_sequences(test_full, WINDOW_SIZE)

print("\nX_train shape:", X_train.shape, "(samples, window_size, features)")
print("y_train shape:", y_train.shape, "(samples, features)")
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
class WorldModel(nn.Module):
    def __init__(self, n_features, hidden_size=64, num_layers=2):
        super().__init__()
        self.lstm = nn.LSTM(n_features, hidden_size, num_layers, batch_first=True)
        self.head = nn.Linear(hidden_size, n_features)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.head(out[:, -1, :])  # predicted next state vector

n_features = X_train.shape[2]
model = WorldModel(n_features).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.MSELoss()

print(model)
```

```python
def weighted_mse(pred, target, attack_weight=5.0):
    weights = torch.ones(pred.shape[1], device=pred.device)
    weights[-1] = attack_weight  # last column is attack_ratio
    return ((pred - target) ** 2 * weights).mean()

model = WorldModel(n_features).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

EPOCHS = 30
for epoch in range(EPOCHS):
    model.train()
    total_loss = 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = weighted_mse(model(xb), yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
    train_loss = total_loss / len(X_train_t)

    model.eval()
    with torch.no_grad():
        test_loss = 0
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            test_loss += weighted_mse(model(xb), yb).item() * xb.size(0)
        test_loss /= len(X_test_t)

    if epoch % 5 == 0 or epoch == EPOCHS - 1:
        print(f"Epoch {epoch:2d} | train wMSE {train_loss:.4f} | test wMSE {test_loss:.4f}")
```

```python
def make_multistep_sequences(data, window_size, horizon):
    X, Y = [], []
    for i in range(len(data) - window_size - horizon + 1):
        X.append(data[i:i+window_size])
        Y.append(data[i+window_size : i+window_size+horizon])
    return np.array(X), np.array(Y)

HORIZON = 5  # train the model to stay accurate 5 steps into its own rollout

X_train_multi, Y_train_multi = make_multistep_sequences(train_full2, WINDOW_SIZE, HORIZON)
X_test_multi, Y_test_multi = make_multistep_sequences(test_full2, WINDOW_SIZE, HORIZON)

print(X_train_multi.shape, Y_train_multi.shape)
print(X_test_multi.shape, Y_test_multi.shape)
```

```python
def train_multistep(model, X, Y, optimizer, horizon, attack_weight=5.0, batch_size=32):
    model.train()
    X_t = torch.tensor(X, dtype=torch.float32)
    Y_t = torch.tensor(Y, dtype=torch.float32)
    loader = DataLoader(TensorDataset(X_t, Y_t), batch_size=batch_size, shuffle=True)

    total_loss = 0
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        window = xb
        loss = 0
        for t in range(horizon):
            pred = model(window)              # predict next state
            target = yb[:, t, :]              # true state at this horizon step
            loss += weighted_mse(pred, target, attack_weight)
            window = torch.cat([window[:, 1:, :], pred.unsqueeze(1)], dim=1)  # feed own prediction back in
        loss = loss / horizon
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
    return total_loss / len(X)


def eval_multistep(model, X, Y, horizon, attack_weight=5.0, batch_size=32):
    model.eval()
    X_t = torch.tensor(X, dtype=torch.float32)
    Y_t = torch.tensor(Y, dtype=torch.float32)
    loader = DataLoader(TensorDataset(X_t, Y_t), batch_size=batch_size, shuffle=False)

    total_loss = 0
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            window = xb
            loss = 0
            for t in range(horizon):
                pred = model(window)
                target = yb[:, t, :]
                loss += weighted_mse(pred, target, attack_weight)
                window = torch.cat([window[:, 1:, :], pred.unsqueeze(1)], dim=1)
            loss = loss / horizon
            total_loss += loss.item() * xb.size(0)
    return total_loss / len(X)

print("Functions defined.")
```

```python
# Cell 29 — Prepare binary attack targets and define classifier

# Convert attack_ratio into binary target:
# 0 = normal, 1 = attack
y_train_attack = (y_train2[:, -1] > 0).astype(np.float32)
y_test_attack = (y_test2[:, -1] > 0).astype(np.float32)

print("y_train_attack shape:", y_train_attack.shape)
print("Training attacks:", int(y_train_attack.sum()))
print("Training normal:", int((y_train_attack == 0).sum()))

# Calculate class weight
pos_weight = torch.tensor(
    [(1 - y_train_attack.mean()) / y_train_attack.mean()],
    dtype=torch.float32
).to(device)

print("pos_weight:", pos_weight.item())


class AttackClassifier(nn.Module):
    def __init__(self, n_features, hidden_size=64, num_layers=2):
        super().__init__()
        self.lstm = nn.LSTM(
            n_features,
            hidden_size,
            num_layers,
            batch_first=True
        )
        self.head = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.head(out[:, -1, :]).squeeze(-1)


print("Classifier defined.")
```

```python
# Cell 30 — Train and evaluate LSTM classifier using the current train/test split

from sklearn.metrics import precision_score, recall_score, f1_score

def train_eval_classifier(X_tr, y_tr, X_te, y_te, seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    # Calculate class weight
    pw = torch.tensor(
        [(1 - y_tr.mean()) / max(y_tr.mean(), 1e-6)],
        dtype=torch.float32
    ).to(device)

    clf = AttackClassifier(n_features).to(device)

    opt = torch.optim.Adam(clf.parameters(), lr=1e-3)
    loss_fn = nn.BCEWithLogitsLoss(pos_weight=pw)

    # Training tensors
    Xtr_t = torch.tensor(X_tr, dtype=torch.float32)
    ytr_t = torch.tensor(y_tr, dtype=torch.float32)

    loader = DataLoader(
        TensorDataset(Xtr_t, ytr_t),
        batch_size=32,
        shuffle=True
    )

    # Train
    for epoch in range(30):
        clf.train()

        for xb, yb in loader:
            xb = xb.to(device)
            yb = yb.to(device)

            opt.zero_grad()

            logits = clf(xb)
            loss = loss_fn(logits, yb)

            loss.backward()
            opt.step()

    # Test
    clf.eval()

    with torch.no_grad():
        test_tensor = torch.tensor(X_te, dtype=torch.float32).to(device)
        logits = clf(test_tensor)
        probs = torch.sigmoid(logits).cpu().numpy()

    preds = (probs > 0.5).astype(int)

    return {
        'precision': precision_score(y_te, preds, zero_division=0),
        'recall': recall_score(y_te, preds, zero_division=0),
        'f1': f1_score(y_te, preds, zero_division=0),
        'test_attack_pct': y_te.mean()
    }


# Use the correctly aligned train/test sequences
result = train_eval_classifier(
    X_train2,
    y_train_attack,
    X_test2_fixed,
    y_test_attack_fixed
)

print("===== LSTM CLASSIFIER RESULT =====")
print(f"Attack percentage: {result['test_attack_pct']:.3f}")
print(f"Precision:          {result['precision']:.3f}")
print(f"Recall:             {result['recall']:.3f}")
print(f"F1 Score:           {result['f1']:.3f}")
```

```python
# Cell 79 — Create sequences without target leakage

def make_feature_target_sequences(features, target, window_size):
    X, y = [], []

    for i in range(len(features) - window_size):
        X.append(features[i:i + window_size])
        y.append(target[i + window_size])

    return np.array(X), np.array(y)


# Training sequences
X_train_no_target, y_train_no_target = make_feature_target_sequences(
    train_scaled_no_target,
    train_df2[['attack_ratio']].values,
    WINDOW_SIZE
)

# Testing sequences:
# Include the last 10 training feature rows as history,
# but predict only testing targets.
combined_features_no_target = np.vstack([
    train_scaled_no_target[-WINDOW_SIZE:],
    test_scaled_no_target
])

combined_targets_no_target = np.vstack([
    train_df2[['attack_ratio']].values[-WINDOW_SIZE:],
    test_df2[['attack_ratio']].values
])

X_test_no_target, y_test_no_target = make_feature_target_sequences(
    combined_features_no_target,
    combined_targets_no_target,
    WINDOW_SIZE
)

print("X_train shape:", X_train_no_target.shape)
print("y_train shape:", y_train_no_target.shape)

print("\nX_test shape:", X_test_no_target.shape)
print("y_test shape:", y_test_no_target.shape)

print("\nInput features per timestep:", X_train_no_target.shape[2])
print("Target values per prediction:", y_train_no_target.shape[1])
```

```python
# Cell 80 — LSTM without attack_ratio as an input

class NetworkBehaviorLSTM(nn.Module):
    def __init__(self, n_features, hidden_size=64, num_layers=2):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=n_features,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )

        self.head = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.head(out[:, -1, :])

n_features_no_target = X_train_no_target.shape[2]

model_no_target = NetworkBehaviorLSTM(
    n_features=n_features_no_target
).to(device)

print(model_no_target)
```

```python
# Cell 115 — Create 3-level risk classification

risk_scores = lr_temporal_probabilities

risk_levels = np.where(
    risk_scores >= 0.50,
    "HIGH",
    np.where(
        risk_scores >= 0.20,
        "MEDIUM",
        "LOW"
    )
)

risk_results = pd.DataFrame({
    "Timestamp": test_times_final,
    "Attack_Probability": risk_scores,
    "Risk_Level": risk_levels,
    "Actual_Attack": y_test_binary_no_target
})

print("===== RISK LEVEL DETECTOR =====")

print(
    risk_results["Risk_Level"]
    .value_counts()
)

print("\nFirst 20 results:")

print(
    risk_results.head(20).to_string(index=False)
)
```

```python
# Cell 116 — Convert risk levels into security alerts

def generate_alert(row):
    if row["Risk_Level"] == "HIGH":
        return "CREATE_INCIDENT"
    elif row["Risk_Level"] == "MEDIUM":
        return "INVESTIGATE"
    else:
        return "NO_ALERT"


risk_results["Action"] = risk_results.apply(
    generate_alert,
    axis=1
)

print("===== AUTOMATIC IR ACTIONS =====")

print(
    risk_results["Action"].value_counts()
)

print("\nSample alerts:")

print(
    risk_results[
        risk_results["Action"] != "NO_ALERT"
    ].head(20).to_string(index=False)
)
```

```python
# Cell 117 — Map detected attack types to MITRE ATT&CK techniques

def map_attack_technique(row):
    if row["Actual_Attack"] == 1:
        # For this CIC-IDS-2018 test period, the attack is FTP-BruteForce
        return "T1110 - Brute Force"
    else:
        return "None"


risk_results["MITRE_Technique"] = risk_results.apply(
    map_attack_technique,
    axis=1
)

print("===== MITRE ATT&CK MAPPING =====")

print(
    risk_results[
        risk_results["Action"] != "NO_ALERT"
    ][
        [
            "Timestamp",
            "Attack_Probability",
            "Risk_Level",
            "Action",
            "MITRE_Technique"
        ]
    ].head(20).to_string(index=False)
)
```
